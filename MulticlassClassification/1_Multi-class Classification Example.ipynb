```python
data = pd.read_csv("https://gist.githubusercontent.com/netj/8836201/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv")

```

```python

#Seperate data as features and label
features = data.iloc[:,[0,1,2,3]].values
label = data.iloc[:,[4]].values
```

```python
# Rules for Classification
# ===========================================================================
# 1. Data must be COMPLETE
# 2. Data must be STRICTLY NUMERIC
# 3. Features and label must be in the form 2d np array
# 4. Scaling of features is optional, however recommended by PN
# 5. For Label (Mandatory)
#        a. Binary Classification - Label must be represented in the form of 0 and 1
#        b. Multi-class Classification - Label must be Discrete Numerical or Dummy Variable
```

# Using Discrete Numerical Label Representation

```python
label
```

```python
#Label Encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
label = le.fit_transform(label)
```

```python
label
```

```python
le.classes_
```

```python
le.transform(np.array([['Virginica']]))
```

```python
le.inverse_transform(np.array([0]))
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
features = sc.fit_transform(features)
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(features,
                                                    label,
                                                    test_size=0.2,
                                                    random_state=10)
```

# Model Building

```python
import tensorflow as tf
```

```python
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=6, activation='relu', input_shape=(4,)))
model.add(tf.keras.layers.Dense(units=6, activation='relu'))
model.add(tf.keras.layers.Dense(units=6, activation='relu'))
model.add(tf.keras.layers.Dense(units=6, activation='relu'))
model.add(tf.keras.layers.Dense(units=3, activation='softmax'))
```

```python
#Compile Model
#Error Functions
# 1. Binary Classification ------ binary_crossentropy
# 2. Multi-class Classification - categorical_crossentropy | sparse_categorical_crossentropy
#
#     if label is discrete numerical:
#              use sparse_categorical_crossentropy
#     else:
#              use categorical_crossentropy
```

```python
model.compile(optimizer="sgd",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
```

```python
class PNCLRuleMonitor(tf.keras.callbacks.Callback):
  def __init__(self,CL):
    super(PNCLRuleMonitor,self).__init__()
    self.CL = CL


  def on_epoch_end(self,epoch,logs=None):
    testScore = logs['val_accuracy']
    trainScore = logs['accuracy']

    if testScore > trainScore and testScore >= self.CL:
      self.model.stop_training = True
```

```python
model.fit(X_train,y_train,
          validation_data=(X_test,y_test),
          epochs=100000,
          callbacks=[PNCLRuleMonitor(0.8)])
```

```python
# Deploy Code

#User input
sl = float(input("Enter s-length: "))
sw = float(input("Enter s-width: "))
pl = float(input("Enter p-length: "))
pw = float(input("Enter p-width: "))

#Convert it into np array
inpuitArray = np.array([[sl,sw,pl,pw]])

#Scale the data
yinputScaled = sc.transform(inpuitArray)

#Predict variety
pred = model.predict(yinputScaled)

predicted_class_index = np.argmax(pred)

print(predicted_class_index)

predicted_variety = le.inverse_transform([predicted_class_index])

print(f"The predicted variety is: {predicted_variety[0]}")
```

# Guidelines for choosing number of units and no of hidden layers in a typical ANN (PN)

```python
# Classification Problem
# ==============================================================================
#
# Option 1: No of hidden layers : no of feature columns
#           No of units per hidden layer: no of feature columns
#
# Option 2: No of hidden layers : no of feature columns
#           No of units per hidden layer: 3 * no of feature columns
#
# Option 3: No of hidden layers : no of feature columns
#           No of units per hidden layer: 1/3 * no of feature columns


# Regression Problem
# ==============================================================================
#
# No of hidden layers
#   1. no of feature cols
#   2. 3 * no of feature cols
#   3. 1/3 * no of feature cols
#
# No of units per hidden layer
#
# Always start with unit: 100 and increment in the multiples of 5
```
