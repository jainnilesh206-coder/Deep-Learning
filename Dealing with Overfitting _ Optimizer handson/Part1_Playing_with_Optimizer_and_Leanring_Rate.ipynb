```python
# Rules for Classification using ANN
# ==============================================
# 1. Data must be complete
# 2. Data must be strictly numeric
# 3. Features and label must be in the form of 2d np array
# 4. Scaling of features is OPTIONAL (But Prashant Nair apply the same rules of Scaling as in Regression)
# 5. For Label (MANDATORY)
#        a. Binary Classification: Label must be represented in the form of 0 or 1
#        b. Multiclass Classification: Label must be either DISCRETE NUMERICAL or DUMMY VARIABLE
```

```python
#Seperate data as features and label
features = data.iloc[:,[0,1]].values
label = data.iloc[:,[2]].values
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
features = sc.fit_transform(features)
```

```python
#Label Normalization
# Here data is by default 0 and 1 , so no changes/preprocessing required
```

```python
#Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(features,
                                                 label,
                                                 test_size=0.2,
                                                 random_state=10)
```

#ANN Modelling

# Changing Optimizer Algo

```python
#Architect the model
import tensorflow as tf

model = tf.keras.Sequential()

model.add(tf.keras.layers.Dense(units = 6 , activation="relu", input_shape=(2,)))
model.add(tf.keras.layers.Dense(units = 6 , activation="relu"))
model.add(tf.keras.layers.Dense(units = 1 , activation="sigmoid"))
```

```python
#Compile Model

#For Classification
# Loss Function:
# a. Binary Classification: binary_crossentropy
# b. Multi-class Classification: categorical_crossentropy | sparse_categorical_crossentropy


model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
class MyCLRuleMonitor(tf.keras.callbacks.Callback):
  def __init__(self,cl):
    super(MyCLRuleMonitor,self).__init__()
    self.cl=cl

  def on_epoch_end(self,epoch,logs=None):
    testScore = logs['val_accuracy']
    trainScore = logs['accuracy']

    if testScore > trainScore and testScore >= self.cl:
      self.model.stop_training=True
```

```python
model.fit(X_train,y_train, validation_data=(X_test,y_test), epochs=10000, callbacks=[MyCLRuleMonitor(0.9)])
```

# Tweak the Learning Rate Parameter

```python
#Architect the model
import tensorflow as tf

model1 = tf.keras.Sequential()

model1.add(tf.keras.layers.Dense(units = 6 , activation="relu", input_shape=(2,)))
model1.add(tf.keras.layers.Dense(units = 6 , activation="relu"))
model1.add(tf.keras.layers.Dense(units = 1 , activation="sigmoid"))
```

```python
#Compile Model

#For Classification
# Loss Function:
# a. Binary Classification: binary_crossentropy
# b. Multi-class Classification: categorical_crossentropy | sparse_categorical_crossentropy


model1.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
               loss="binary_crossentropy",
               metrics=["accuracy"])
```

```python
class MyCLRuleMonitor(tf.keras.callbacks.Callback):
  def __init__(self,cl):
    super(MyCLRuleMonitor,self).__init__()
    self.cl=cl

  def on_epoch_end(self,epoch,logs=None):
    testScore = logs.get('val_accuracy')
    trainScore = logs.get('accuracy')

    if testScore > trainScore and testScore >= self.cl:
      self.model.stop_training=True
```

```python
model1.fit(X_train,y_train, validation_data=(X_test,y_test), epochs=10000, callbacks=[MyCLRuleMonitor(0.9)])
```
